```python
from gtts import gTTS
import os
```

```python
def text_to_speech_and_save(text, filename="output.mp3", lang='en'):
    """
    Converts text to speech using gTTS and saves the audio to a local file.

    Args:
        text (str): The text to convert.
        filename (str): The name of the audio file to save.
        lang (str): The language of the text (e.g., 'en' for English, 'es' for Spanish).
    """
    try:
        print(f"Converting text to speech: '{text}'...")
        tts = gTTS(text=text, lang=lang, slow=False)
        tts.save(filename)
        print(f"Audio saved successfully to '{filename}'")
        # Optional: Play the audio immediately after saving
        # os.system(f"start {filename}") # For Windows
        os.system(f"afplay {filename}") # For macOS
        os.system(f"mpg321 {filename}") # For Linux (requires mpg321 to be installed)
    except Exception as e:
        print(f"An error occurred: {e}")

```

```python
# Example usage:
sample_text = "Hello, this is a test of the text-to-speech functionality in Google Colab."
output_audio_file = "my_speech.mp3"
text_to_speech_and_save(sample_text, output_audio_file)

# You can also try with a different language, for example Spanish:
# spanish_text = "Hola, esto es una prueba de la función de texto a voz."
# text_to_speech_and_save(spanish_text, "spanish_speech.mp3", lang='es')
```
